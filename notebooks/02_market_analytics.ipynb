```python
# ============================================================
# Tata Power EZ Charge
# Market & Network Analytics
#
# Project:
# Optimizing EV Charging Utilization Through Demand Shifting
#
# Phase 4 — Exploratory / Market Analytics
# ============================================================


# ============================================================
# 1. Imports
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path


# ============================================================
# 2. Project paths
# ============================================================

DATA_PATH = Path("../data/raw")
OUTPUT_FIGURES = Path("../outputs/figures")
OUTPUT_TABLES = Path("../outputs/tables")

OUTPUT_FIGURES.mkdir(parents=True, exist_ok=True)
OUTPUT_TABLES.mkdir(parents=True, exist_ok=True)


# ============================================================
# 3. Load datasets
# ============================================================

stations = pd.read_csv(DATA_PATH / "stations.csv")
chargers = pd.read_csv(DATA_PATH / "chargers.csv")
customers = pd.read_csv(DATA_PATH / "customers.csv")
monthly_sessions = pd.read_csv(DATA_PATH / "monthly_sessions.csv")
sessions = pd.read_csv(DATA_PATH / "sessions.csv")


# ============================================================
# 4. Parse timestamps
# ============================================================

sessions["start_timestamp"] = pd.to_datetime(
    sessions["start_timestamp"]
)


# ============================================================
# 5. Basic dataset validation
# ============================================================

print("=" * 60)
print("DATASET VALIDATION")
print("=" * 60)

print(f"Stations:          {len(stations):,}")
print(f"Chargers:          {len(chargers):,}")
print(f"Customers:         {len(customers):,}")
print(f"Monthly demand:    {len(monthly_sessions):,}")
print(f"Charging sessions: {len(sessions):,}")

print("\nSession columns:")
print(sessions.columns.tolist())

print("\nMissing values:")
print(sessions.isna().sum())

print("\nDuplicate sessions:")
print(sessions["session_id"].duplicated().sum())

print("\nDate range:")
print(
    sessions["start_timestamp"].min(),
    "to",
    sessions["start_timestamp"].max()
)


# ============================================================
# 6. Create analytical session features
# ============================================================

sessions["date"] = sessions["start_timestamp"].dt.date
sessions["month"] = sessions["start_timestamp"].dt.month
sessions["month_name"] = sessions["start_timestamp"].dt.strftime("%b")

sessions["day_of_week"] = (
    sessions["start_timestamp"]
    .dt.day_name()
)

sessions["hour"] = (
    sessions["start_timestamp"]
    .dt.hour
)

sessions["is_weekend"] = (
    sessions["start_timestamp"]
    .dt.dayofweek >= 5
)


# Peak definition:
# Evening charging demand is treated as the primary
# network-congestion window for baseline analysis.

sessions["period"] = np.where(
    sessions["hour"].between(17, 21),
    "Peak",
    "Off-Peak"
)


# ============================================================
# 7. Overall network KPIs
# ============================================================

completed_sessions = sessions[
    sessions["session_status"] == "Completed"
].copy()

total_sessions = len(sessions)
completed_count = len(completed_sessions)

total_energy = completed_sessions["energy_kwh"].sum()

average_energy = completed_sessions["energy_kwh"].mean()

average_duration = completed_sessions[
    "duration_minutes"
].mean()

completion_rate = (
    completed_count / total_sessions
)

average_price = completed_sessions[
    "base_price_per_kwh"
].mean()

estimated_revenue = (
    completed_sessions["energy_kwh"]
    * completed_sessions["base_price_per_kwh"]
).sum()


network_kpis = pd.Series({
    "Total sessions": total_sessions,
    "Completed sessions": completed_count,
    "Completion rate": completion_rate,
    "Total energy delivered (kWh)": total_energy,
    "Average energy/session (kWh)": average_energy,
    "Average session duration (min)": average_duration,
    "Average price/kWh": average_price,
    "Estimated gross revenue": estimated_revenue
})


print("\n" + "=" * 60)
print("NETWORK KPI SUMMARY")
print("=" * 60)

print(network_kpis)


# ============================================================
# 8. Station network distribution
# ============================================================

stations_by_city = (
    stations
    .groupby("city")
    .size()
    .sort_values(ascending=False)
)

print("\n" + "=" * 60)
print("STATIONS BY CITY")
print("=" * 60)

print(stations_by_city)


plt.figure(figsize=(10, 6))

stations_by_city.sort_values().plot(
    kind="barh"
)

plt.title("Charging Stations by City")
plt.xlabel("Number of Stations")
plt.ylabel("City")
plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "stations_by_city.png",
    dpi=150
)

plt.show()


# ============================================================
# 9. Station type distribution
# ============================================================

stations_by_type = (
    stations["station_type"]
    .value_counts()
    .sort_values(ascending=False)
)

print("\n" + "=" * 60)
print("STATIONS BY TYPE")
print("=" * 60)

print(stations_by_type)


plt.figure(figsize=(10, 6))

stations_by_type.plot(
    kind="bar"
)

plt.title("Charging Stations by Station Type")
plt.xlabel("Station Type")
plt.ylabel("Number of Stations")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "stations_by_type.png",
    dpi=150
)

plt.show()


# ============================================================
# 10. Charger mix
# ============================================================

chargers_by_type = (
    chargers["charger_type"]
    .value_counts()
)

print("\n" + "=" * 60)
print("CHARGER MIX")
print("=" * 60)

print(chargers_by_type)


plt.figure(figsize=(8, 5))

chargers_by_type.plot(
    kind="bar"
)

plt.title("Charger Mix Across the Network")
plt.xlabel("Charger Type")
plt.ylabel("Number of Chargers")

plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "charger_mix.png",
    dpi=150
)

plt.show()


# ============================================================
# 11. Charger capacity by city
# ============================================================

charger_city_summary = (
    chargers
    .merge(
        stations[
            ["station_id", "city", "station_type"]
        ],
        on="station_id",
        how="left"
    )
    .groupby("city")
    .agg(
        chargers=("charger_id", "count"),
        total_power_kw=("power_kw", "sum"),
        avg_power_kw=("power_kw", "mean")
    )
    .sort_values(
        "total_power_kw",
        ascending=False
    )
)

print("\n" + "=" * 60)
print("CHARGING CAPACITY BY CITY")
print("=" * 60)

print(charger_city_summary.round(2))


# ============================================================
# 12. Sessions by city
# ============================================================

city_sessions = (
    sessions
    .groupby("city")
    .agg(
        sessions=("session_id", "count"),
        completed_sessions=(
            "session_status",
            lambda x: (x == "Completed").sum()
        ),
        energy_kwh=(
            "energy_kwh",
            lambda x: x[sessions.loc[x.index, "session_status"] == "Completed"].sum()
        )
    )
    .sort_values(
        "sessions",
        ascending=False
    )
)

city_sessions["completion_rate"] = (
    city_sessions["completed_sessions"]
    / city_sessions["sessions"]
)

city_sessions["energy_mwh"] = (
    city_sessions["energy_kwh"] / 1000
)

print("\n" + "=" * 60)
print("DEMAND BY CITY")
print("=" * 60)

print(city_sessions.round(2))


plt.figure(figsize=(10, 6))

city_sessions["sessions"].sort_values().plot(
    kind="barh"
)

plt.title("Charging Sessions by City")
plt.xlabel("Number of Sessions")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "sessions_by_city.png",
    dpi=150
)

plt.show()


# ============================================================
# 13. Energy delivered by city
# ============================================================

plt.figure(figsize=(10, 6))

city_sessions["energy_mwh"].sort_values().plot(
    kind="barh"
)

plt.title("Energy Delivered by City")
plt.xlabel("Energy Delivered (MWh)")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "energy_by_city.png",
    dpi=150
)

plt.show()


# ============================================================
# 14. Monthly demand
# ============================================================

monthly_demand = (
    sessions
    .groupby("month")
    .agg(
        sessions=("session_id", "count"),
        completed_sessions=(
            "session_status",
            lambda x: (x == "Completed").sum()
        ),
        energy_kwh=(
            "energy_kwh",
            lambda x: x[sessions.loc[x.index, "session_status"] == "Completed"].sum()
        )
    )
)

monthly_demand["energy_mwh"] = (
    monthly_demand["energy_kwh"] / 1000
)

monthly_demand["month_name"] = (
    pd.to_datetime(
        monthly_demand.index,
        format="%m"
    )
    .strftime("%b")
)

print("\n" + "=" * 60)
print("MONTHLY DEMAND")
print("=" * 60)

print(
    monthly_demand[
        [
            "month_name",
            "sessions",
            "completed_sessions",
            "energy_mwh"
        ]
    ].round(2)
)


plt.figure(figsize=(10, 6))

plt.plot(
    monthly_demand.index,
    monthly_demand["sessions"],
    marker="o"
)

plt.title("Monthly Charging Demand")
plt.xlabel("Month")
plt.ylabel("Charging Sessions")

plt.xticks(
    monthly_demand.index,
    monthly_demand["month_name"]
)

plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "monthly_demand.png",
    dpi=150
)

plt.show()


# ============================================================
# 15. Monthly energy delivered
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    monthly_demand.index,
    monthly_demand["energy_mwh"],
    marker="o"
)

plt.title("Monthly Energy Delivered")
plt.xlabel("Month")
plt.ylabel("Energy (MWh)")

plt.xticks(
    monthly_demand.index,
    monthly_demand["month_name"]
)

plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "monthly_energy.png",
    dpi=150
)

plt.show()


# ============================================================
# 16. Hour-of-day demand
# ============================================================

hourly_demand = (
    sessions
    .groupby("hour")
    .size()
)

print("\n" + "=" * 60)
print("HOURLY DEMAND")
print("=" * 60)

print(hourly_demand)


plt.figure(figsize=(11, 5))

plt.plot(
    hourly_demand.index,
    hourly_demand.values,
    marker="o"
)

plt.title("Charging Demand by Hour of Day")
plt.xlabel("Hour")
plt.ylabel("Charging Sessions")

plt.xticks(range(24))

plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "hourly_demand.png",
    dpi=150
)

plt.show()


# ============================================================
# 17. Peak vs off-peak demand
# ============================================================

period_summary = (
    sessions
    .groupby("period")
    .agg(
        sessions=("session_id", "count"),
        completed_sessions=(
            "session_status",
            lambda x: (x == "Completed").sum()
        ),
        energy_kwh=(
            "energy_kwh",
            lambda x: x[sessions.loc[x.index, "session_status"] == "Completed"].sum()
        )
    )
)

period_summary["share_of_sessions"] = (
    period_summary["sessions"]
    / period_summary["sessions"].sum()
)

period_summary["share_of_energy"] = (
    period_summary["energy_kwh"]
    / period_summary["energy_kwh"].sum()
)

print("\n" + "=" * 60)
print("PEAK VS OFF-PEAK")
print("=" * 60)

print(period_summary.round(3))


# ============================================================
# 18. Weekday vs weekend behaviour
# ============================================================

weekday_weekend = (
    sessions
    .groupby("is_weekend")
    .agg(
        sessions=("session_id", "count"),
        energy_kwh=("energy_kwh", "sum"),
        avg_energy=("energy_kwh", "mean"),
        avg_duration=("duration_minutes", "mean")
    )
)

weekday_weekend.index = [
    "Weekday" if x is False else "Weekend"
    for x in weekday_weekend.index
]

print("\n" + "=" * 60)
print("WEEKDAY VS WEEKEND")
print("=" * 60)

print(weekday_weekend.round(2))


plt.figure(figsize=(7, 5))

weekday_weekend["sessions"].plot(
    kind="bar"
)

plt.title("Charging Sessions: Weekday vs Weekend")
plt.xlabel("")
plt.ylabel("Sessions")
plt.xticks(rotation=0)

plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "weekday_weekend_demand.png",
    dpi=150
)

plt.show()


# ============================================================
# 19. Demand by day of week
# ============================================================

day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_demand = (
    sessions
    .groupby("day_of_week")
    .size()
    .reindex(day_order)
)

print("\n" + "=" * 60)
print("DEMAND BY DAY OF WEEK")
print("=" * 60)

print(day_demand)


plt.figure(figsize=(9, 5))

day_demand.plot(
    kind="bar"
)

plt.title("Charging Demand by Day of Week")
plt.xlabel("")
plt.ylabel("Sessions")
plt.xticks(rotation=30)

plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "demand_by_day_of_week.png",
    dpi=150
)

plt.show()


# ============================================================
# 20. Station-level demand
# ============================================================

station_demand = (
    sessions
    .groupby("station_id")
    .agg(
        sessions=("session_id", "count"),
        completed_sessions=(
            "session_status",
            lambda x: (x == "Completed").sum()
        ),
        energy_kwh=(
            "energy_kwh",
            lambda x: x[sessions.loc[x.index, "session_status"] == "Completed"].sum()
        ),
        avg_energy_kwh=("energy_kwh", "mean"),
        avg_duration=("duration_minutes", "mean")
    )
)

station_demand["completion_rate"] = (
    station_demand["completed_sessions"]
    / station_demand["sessions"]
)


station_demand = (
    station_demand
    .merge(
        stations[
            [
                "station_id",
                "city",
                "station_type",
                "charger_count",
                "avg_daily_capacity_kwh",
                "baseline_utilization",
                "peak_demand_factor",
                "operational_reliability"
            ]
        ],
        on="station_id",
        how="left"
    )
)


# ============================================================
# 21. Station performance
# ============================================================

station_demand["sessions_per_charger"] = (
    station_demand["sessions"]
    / station_demand["charger_count"]
)

station_demand["energy_per_charger"] = (
    station_demand["energy_kwh"]
    / station_demand["charger_count"]
)


print("\n" + "=" * 60)
print("TOP 15 STATIONS BY SESSION VOLUME")
print("=" * 60)

print(
    station_demand
    .sort_values("sessions", ascending=False)
    .head(15)
    [
        [
            "station_id",
            "city",
            "station_type",
            "charger_count",
            "sessions",
            "energy_kwh",
            "completion_rate"
        ]
    ]
    .round(2)
)


print("\n" + "=" * 60)
print("BOTTOM 15 STATIONS BY SESSION VOLUME")
print("=" * 60)

print(
    station_demand
    .sort_values("sessions")
    .head(15)
    [
        [
            "station_id",
            "city",
            "station_type",
            "charger_count",
            "sessions",
            "energy_kwh",
            "completion_rate"
        ]
    ]
    .round(2)
)


# ============================================================
# 22. Station type performance
# ============================================================

station_type_performance = (
    station_demand
    .groupby("station_type")
    .agg(
        stations=("station_id", "count"),
        chargers=("charger_count", "sum"),
        sessions=("sessions", "sum"),
        energy_kwh=("energy_kwh", "sum"),
        avg_sessions_per_station=("sessions", "mean"),
        avg_energy_per_station=("energy_kwh", "mean"),
        avg_completion_rate=("completion_rate", "mean")
    )
    .sort_values(
        "sessions",
        ascending=False
    )
)

print("\n" + "=" * 60)
print("PERFORMANCE BY STATION TYPE")
print("=" * 60)

print(station_type_performance.round(2))


# ============================================================
# 23. Charger type performance
# ============================================================

charger_session_performance = (
    sessions
    .groupby("charger_type")
    .agg(
        sessions=("session_id", "count"),
        energy_kwh=("energy_kwh", "sum"),
        avg_energy_kwh=("energy_kwh", "mean"),
        avg_duration=("duration_minutes", "mean"),
        completed_sessions=(
            "session_status",
            lambda x: (x == "Completed").sum()
        )
    )
)

charger_session_performance["completion_rate"] = (
    charger_session_performance["completed_sessions"]
    / charger_session_performance["sessions"]
)

print("\n" + "=" * 60)
print("PERFORMANCE BY CHARGER TYPE")
print("=" * 60)

print(charger_session_performance.round(2))


plt.figure(figsize=(8, 5))

charger_session_performance["sessions"].plot(
    kind="bar"
)

plt.title("Sessions by Charger Type")
plt.xlabel("Charger Type")
plt.ylabel("Sessions")
plt.xticks(rotation=0)

plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "sessions_by_charger_type.png",
    dpi=150
)

plt.show()


# ============================================================
# 24. Session duration distribution
# ============================================================

print("\n" + "=" * 60)
print("SESSION DURATION")
print("=" * 60)

print(
    completed_sessions[
        "duration_minutes"
    ]
    .describe()
    .round(2)
)


plt.figure(figsize=(9, 5))

plt.hist(
    completed_sessions["duration_minutes"],
    bins=40
)

plt.title("Distribution of Charging Session Duration")
plt.xlabel("Duration (minutes)")
plt.ylabel("Sessions")

plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "session_duration_distribution.png",
    dpi=150
)

plt.show()


# ============================================================
# 25. Energy per session distribution
# ============================================================

print("\n" + "=" * 60)
print("ENERGY PER SESSION")
print("=" * 60)

print(
    completed_sessions[
        "energy_kwh"
    ]
    .describe()
    .round(2)
)


plt.figure(figsize=(9, 5))

plt.hist(
    completed_sessions["energy_kwh"],
    bins=40
)

plt.title("Distribution of Energy Delivered per Session")
plt.xlabel("Energy (kWh)")
plt.ylabel("Sessions")

plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "energy_distribution.png",
    dpi=150
)

plt.show()


# ============================================================
# 26. Session status analysis
# ============================================================

status_summary = (
    sessions["session_status"]
    .value_counts()
)

status_share = (
    status_summary
    / status_summary.sum()
)

status_table = pd.DataFrame({
    "sessions": status_summary,
    "share": status_share
})

print("\n" + "=" * 60)
print("SESSION STATUS")
print("=" * 60)

print(status_table.round(4))


plt.figure(figsize=(7, 5))

status_summary.plot(
    kind="bar"
)

plt.title("Charging Session Outcomes")
plt.xlabel("Session Status")
plt.ylabel("Sessions")
plt.xticks(rotation=0)

plt.tight_layout()

plt.savefig(
    OUTPUT_FIGURES / "session_status.png",
    dpi=150
)

plt.show()


# ============================================================
# 27. Failed and cancelled sessions by city
# ============================================================

problem_sessions = sessions[
    sessions["session_status"] != "Completed"
]

problem_by_city = (
    problem_sessions
    .groupby("city")
    .size()
    .sort_values(ascending=False)
)

print("\n" + "=" * 60)
print("FAILED + CANCELLED SESSIONS BY CITY")
print("=" * 60)

print(problem_by_city)


# ============================================================
# 28. Utilization opportunity analysis
# ============================================================

# Station-level observed demand is compared against
# the station's baseline utilization assumption.

utilization_opportunity = station_demand[
    [
        "station_id",
        "city",
        "station_type",
        "charger_count",
        "sessions",
        "energy_kwh",
        "baseline_utilization",
        "peak_demand_factor",
        "operational_reliability"
    ]
].copy()


# Lower baseline utilization = potentially greater
# opportunity to absorb additional demand.

utilization_opportunity["opportunity_score"] = (
    1
    - utilization_opportunity["baseline_utilization"]
)


print("\n" + "=" * 60)
print("POTENTIAL CAPACITY OPPORTUNITIES")
print("=" * 60)

print(
    utilization_opportunity
    .sort_values(
        "opportunity_score",
        ascending=False
    )
    .head(15)
    .round(3)
)


# ============================================================
# 29. City-level network opportunity
# ============================================================

city_capacity = (
    stations
    .groupby("city")
    .agg(
        stations=("station_id", "count"),
        chargers=("charger_count", "sum"),
        avg_baseline_utilization=(
            "baseline_utilization",
            "mean"
        ),
        avg_operational_reliability=(
            "operational_reliability",
            "mean"
        ),
        avg_peak_demand_factor=(
            "peak_demand_factor",
            "mean"
        )
    )
)

city_analysis = (
    city_capacity
    .join(
        city_sessions[
            [
                "sessions",
                "energy_kwh",
                "completion_rate"
            ]
        ],
        how="left"
    )
)

city_analysis["sessions_per_charger"] = (
    city_analysis["sessions"]
    / city_analysis["chargers"]
)

city_analysis["energy_per_charger"] = (
    city_analysis["energy_kwh"]
    / city_analysis["chargers"]
)

print("\n" + "=" * 60)
print("CITY-LEVEL NETWORK ANALYSIS")
print("=" * 60)

print(
    city_analysis
    .sort_values(
        "sessions_per_charger",
        ascending=False
    )
    .round(2)
)


# ============================================================
# 30. Peak demand by city
# ============================================================

peak_city = (
    sessions[
        sessions["period"] == "Peak"
    ]
    .groupby("city")
    .size()
    .sort_values(ascending=False)
)

print("\n" + "=" * 60)
print("PEAK DEMAND BY CITY")
print("=" * 60)

print(peak_city)


# ============================================================
# 31. Peak share by city
# ============================================================

city_period = pd.crosstab(
    sessions["city"],
    sessions["period"],
    normalize="index"
)

city_period["Peak_share"] = (
    city_period.get("Peak", 0)
)

city_period = (
    city_period
    .sort_values(
        "Peak_share",
        ascending=False
    )
)

print("\n" + "=" * 60)
print("PEAK DEMAND SHARE BY CITY")
print("=" * 60)

print(city_period.round(3))


# ============================================================
# 32. Peak demand by station type
# ============================================================

station_period = (
    sessions
    .groupby(
        ["station_type", "period"]
    )
    .size()
    .unstack(fill_value=0)
)

station_period["peak_share"] = (
    station_period["Peak"]
    / station_period.sum(axis=1)
)

print("\n" + "=" * 60)
print("PEAK DEMAND SHARE BY STATION TYPE")
print("=" * 60)

print(station_period.round(3))


# ============================================================
# 33. Customer segment baseline behaviour
# ============================================================

customer_segment_sessions = (
    sessions
    .merge(
        customers[
            [
                "customer_id",
                "customer_segment",
                "base_sessions_per_month",
                "peak_preference_score",
                "price_sensitivity_score",
                "convenience_score",
                "incentive_response_score",
                "home_charging_access"
            ]
        ],
        on="customer_id",
        how="left"
    )
)


segment_baseline = (
    customer_segment_sessions
    .groupby("customer_segment")
    .agg(
        customers=("customer_id", "nunique"),
        sessions=("session_id", "count"),
        avg_energy_kwh=("energy_kwh", "mean"),
        avg_duration=("duration_minutes", "mean"),
        peak_sessions=(
            "period",
            lambda x: (x == "Peak").sum()
        ),
        completed_sessions=(
            "session_status",
            lambda x: (x == "Completed").sum()
        ),
        avg_peak_preference=(
            "peak_preference_score",
            "mean"
        ),
        avg_price_sensitivity=(
            "price_sensitivity_score",
            "mean"
        ),
        avg_incentive_response=(
            "incentive_response_score",
            "mean"
        )
    )
)

segment_baseline["peak_share"] = (
    segment_baseline["peak_sessions"]
    / segment_baseline["sessions"]
)

segment_baseline["completion_rate"] = (
    segment_baseline["completed_sessions"]
    / segment_baseline["sessions"]
)

print("\n" + "=" * 60)
print("CUSTOMER SEGMENT BASELINE BEHAVIOUR")
print("=" * 60)

print(
    segment_baseline.round(3)
)


# ============================================================
# 34. Save important analytical tables
# ============================================================

city_sessions.to_csv(
    OUTPUT_TABLES / "city_sessions.csv"
)

city_analysis.to_csv(
    OUTPUT_TABLES / "city_network_analysis.csv"
)

station_demand.to_csv(
    OUTPUT_TABLES / "station_performance.csv",
    index=False
)

station_type_performance.to_csv(
    OUTPUT_TABLES / "station_type_performance.csv"
)

charger_session_performance.to_csv(
    OUTPUT_TABLES / "charger_type_performance.csv"
)

segment_baseline.to_csv(
    OUTPUT_TABLES / "segment_baseline.csv"
)

period_summary.to_csv(
    OUTPUT_TABLES / "peak_offpeak_summary.csv"
)

monthly_demand.to_csv(
    OUTPUT_TABLES / "monthly_demand.csv"
)

status_table.to_csv(
    OUTPUT_TABLES / "session_status_summary.csv"
)


# ============================================================
# 35. Baseline analytical observations
# ============================================================

print("\n" + "=" * 60)
print("BASELINE ANALYTICAL OBSERVATIONS")
print("=" * 60)


top_city = city_sessions["sessions"].idxmax()
top_city_sessions = city_sessions.loc[
    top_city,
    "sessions"
]

highest_peak_city = peak_city.idxmax()

highest_peak_share_city = (
    city_period["Peak_share"].idxmax()
)

highest_demand_station_type = (
    station_type_performance["sessions"].idxmax()
)

highest_demand_charger_type = (
    charger_session_performance["sessions"].idxmax()
)


print(
    f"\n1. Highest session-volume city: "
    f"{top_city} "
    f"({top_city_sessions:,.0f} sessions)"
)

print(
    f"\n2. City with highest absolute peak demand: "
    f"{highest_peak_city}"
)

print(
    f"\n3. City with highest peak-session share: "
    f"{highest_peak_share_city} "
    f"({city_period.loc[highest_peak_share_city, 'Peak_share']:.1%})"
)

print(
    f"\n4. Highest-demand station type: "
    f"{highest_demand_station_type}"
)

print(
    f"\n5. Highest-demand charger type: "
    f"{highest_demand_charger_type}"
)

print(
    f"\n6. Peak sessions account for: "
    f"{period_summary.loc['Peak', 'share_of_sessions']:.1%} "
    f"of all sessions"
)

print(
    f"\n7. Off-peak sessions account for: "
    f"{period_summary.loc['Off-Peak', 'share_of_sessions']:.1%} "
    f"of all sessions"
)


# ============================================================
# 36. Final interpretation
# ============================================================

print("\n" + "=" * 60)
print("PHASE 4 COMPLETE")
print("=" * 60)

print(
    """
The network baseline has now been established.

The next analytical questions are:

1. Which customers are most dependent on peak charging?
2. Which customers appear capable of shifting their charging time?
3. Can we construct a defensible Charging Flexibility Score?
4. Which customer segments should be considered eligible
   for a targeted intervention?

These questions will form the next phase:
Customer Behaviour & Segmentation.
"""
)
```
